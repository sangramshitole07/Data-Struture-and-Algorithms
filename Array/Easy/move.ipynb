```python
#move zero to right 

class Sol:
    
    def moveZero(self,nums):
        j=0
        
        for i in range(len(nums)):
            if nums[i]!=0:
                nums[j]=nums[i]
                j+=1
        
        while j < len(nums):
            nums[j]=0
            j+=1
        
if __name__=="__main__":
    sol=Sol()
    nums=[1,0,3,4,0,2,0,1]
    sol.moveZero(nums)
    print(nums)
```
