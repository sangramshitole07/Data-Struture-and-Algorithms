```python
arr=[1,1,1,0,0,1,1,1,1,0]

count=0
max_count=0

for i in arr:
    if i==1:
        count+=1
        if count > max_count:
            max_count=count
    
    else:
        count=0
    
print(max_count)
```

```python
def max_con_one(arr):
    count=0
    max_count=0
    
    for i in arr:
        if i ==1:
            count+=1
            if count>max_count:
                max_count=count
        else:
            count=0
    
    return max_count

arr=[1,2,0,0,1]
arr1=[2,3,4,8]
result=max_con_one(arr)
print(result)
res2=max_con_one(arr1)
print(res2)
```
