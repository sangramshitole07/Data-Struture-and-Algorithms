```python
class Solu:
    def longest_Subarray(self, nums, k):
        maxLen = 0
        n = len(nums)

        for start in range(n):

            sum = 0

            for end in range(start, n):

                sum += nums[end]

                if sum == k:
                    maxLen = max(maxLen, end - start + 1)

        return maxLen


if __name__ == "__main__":
    sol = Solu()

    k = 6
    nums = [2, 3, 2, 1, 1, 1]

    res = sol.longest_Subarray(nums, k)

    print(res)
```
