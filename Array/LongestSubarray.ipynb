```python
class Solu:
    def longest_Subarray(self, nums, k):
        maxLen = 0
        n = len(nums)

        for start in range(n):

            sum = 0

            for end in range(start, n):

                sum += nums[end]

                if sum == k:
                    maxLen = max(maxLen, end - start + 1)

        return maxLen


if __name__ == "__main__":
    sol = Solu()

    k = 6
    nums = [2, 3, 2, 1, 1, 1]

    res = sol.longest_Subarray(nums, k)

    print(res)
```

```python
class Sol:
    def longestSubarray(self,nums,k):
        n=len(nums)
        maxlen=0
        left=0
        right=0
        sum=nums[0]
        
        while right<n:
            while left <=right and sum > k:
                sum-=nums[left]
                left+=1
            
            if sum==k:
                maxlen=max(maxlen,right-left+1)
            
            right+=1
            if right <n:
                sum+=nums[right]
        return maxlen

nums=[10,5,2,7,1,9]
k=15
sol=Sol()
ans=sol.longestSubarray(nums,k)
print(" Length of longest subarray having sum k is : ", ans)
                
```
