```python
class Solution:
    def singleNumber(self, nums):
        
        for num in nums:
            count = 0

            for x in nums:
                if x == num:
                    count += 1

            if count == 1:
                return num

        return -1


nums=[1,2,2]
obj=Solution()

res=obj.singleNumber(nums)
print(res)
```

```python
class Solution:
    def singleNumber(self, nums):
        
        xorr = 0

        # nums = [4, 1, 2, 1, 2]
        #
        # Initially:
        # xorr = 0
        #
        # -------------------------
        # 1st iteration:
        # num = 4
        # xorr = 0 ^ 4
        #      = 4
        #
        # xorr = 4
        #
        # -------------------------
        # 2nd iteration:
        # num = 1
        # xorr = 4 ^ 1
        #      = 5
        #
        # xorr = 5
        #
        # -------------------------
        # 3rd iteration:
        # num = 2
        # xorr = 5 ^ 2
        #      = 7
        #
        # xorr = 7
        #
        # -------------------------
        # 4th iteration:
        # num = 1
        # xorr = 7 ^ 1
        #      = 6
        #
        # xorr = 6
        #
        # -------------------------
        # 5th iteration:
        # num = 2
        # xorr = 6 ^ 2
        #      = 4
        #
        # xorr = 4
        #
        # -------------------------
        # Final answer:
        # xorr = 4
        #
        # Therefore, the number that appears only once is 4.

        for num in nums:
            xorr ^= num

        return xorr


nums = [4, 1, 2, 1, 2]

obj = Solution()
result = obj.singleNumber(nums)

print(result)
```
