```python
#implementing linear search

class Linear():
    
    def search(self,arr,x):
        if arr==None:
            return -1
        for i in range(0,len(arr)):
            
            if arr[i]==x:
                return i
        
    
        return 0

if __name__=="__main__":
    arr=[1,2,3,45,6,6,3,42,32,32134,32,4,3,2]
    x=3
    
    Ln=Linear()
    result=Ln.search(arr,x)
    print("The element is preseat at index :", result)
    
```
