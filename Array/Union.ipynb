```python
#union of 2 sorted array using set

class Solution:
    def unionOfSorted(self,arr1,arr2,a1,a2):
        freq={}
        
        for i in range(a1):
            freq[arr1[i]]=freq.get(arr1[i],0)+1
            
        
        for i in range(a2):
            freq[arr2[i]]=freq.get(arr2[i],0)+1
            
        union=sorted(freq.keys())
        
        return union
    
if __name__=="__main__":
    arr1=[1,2,3,3,4,4,5]
    arr2=[2,2,3,4,4,4,5,6]
    
    a1=len(arr1)
    a2=len(arr2)
    
    sol=Solution()
    result=sol.unionOfSorted(arr1,arr2,a1,a2)
    print("Union of arr1 and arr2 is : ", result)
                    
```

```python
#union of two arrays using the union of property 

class Solution:
    def findUnion(self,arr1,arr2):
        st= set(arr1) | set(arr2)
        
        return sorted(st)

if __name__=="__main__":
    arr1=[1,2,3,4,5,6,7,7,8,8,9,23]
    arr2=[2,3,4,5,6,6,6,7]
    obj=Solution()
    result=obj.findUnion(arr1,arr2)
    print("union of arr1 and arr2 is : " ,result)
```

```python

# using two pointer 
class Solution:

    def findUnion(self, arr1, arr2):
        Union = []
        i = j = 0

        while i < len(arr1) and j < len(arr2):

            if arr1[i] < arr2[j]:
                x = arr1[i]
                i += 1

            elif arr2[j] < arr1[i]:
                x = arr2[j]
                j += 1

            else:
                x = arr1[i]
                i += 1
                j += 1

            if not Union or Union[-1] != x:
                Union.append(x)

        while i < len(arr1):
            #If the list is empty OR the last element is different from x, add x.
            if not Union or Union[-1] != arr1[i]:
                Union.append(arr1[i])
            i += 1

        while j < len(arr2):
            if not Union or Union[-1] != arr2[j]:
                Union.append(arr2[j])
            j += 1

        return Union
    
if __name__=="__main__":
    arr1=[1,2,3,4,4,4,4,4,5]
    arr2=[4,5,6,7]
    sol=Solution()
    result=sol.findUnion(arr1,arr2)
    print("Union of the 2 arrays is : ",result)
```
