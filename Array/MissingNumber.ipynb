```python
#nlog n time complexity for finding the missing element

def missingNumber(arr):
        arr.sort()
        
        expected=1
        for i in range(len(arr)):
            if arr[i]!=expected:
                return expected
        
            expected+=1
        return expected
arr=[6,2,3,4,5,7,1]
print(missingNumber(arr))
        
            
```
