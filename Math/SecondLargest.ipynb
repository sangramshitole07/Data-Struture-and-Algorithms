```python
def Optimal(arr, n):
    if n < 2:
        return -1

    small = float('inf')
    second = float('inf')

    for i in range(n):
        if arr[i] < small:
            second = small
            small = arr[i]
        elif arr[i] < second and arr[i] != small:
            second = arr[i]

    return second


if __name__ == "__main__":
    arr = [12, 3, 23, 21, 12, 4, 3, 432.33]
    n = len(arr)
    second_smallest = Optimal(arr, n)
    print("Value of second Small element is : ", second_smallest)
```
