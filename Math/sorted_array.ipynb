```python
#brute force
def Is_Sorted(arr,n):
    for i in range(n):
        for j in range(i + 1,n):
            if arr[j] < arr[i]:
                return False       
    return True


    
arr=[1,2,3,6,5,92,33]
n=len(arr)
ans = Is_Sorted(arr,n)
if  ans:
    print("true")
else:
    print("false")
   
```
