Sort an array of 0s, 1s and 2s



Problem Statement: Given an array nums consisting of only 0, 1, or 2. Sort the array in non-decreasing order. The sorting must be done in-place, without making a copy of the original array.

Examples
Input: nums = [1, 0, 2, 1, 0]
Output: [0, 0, 1, 1, 2]
Explanation: The nums array in sorted order has 2 zeroes, 2 ones and 1 two

Input: nums = [0, 0, 1, 1, 1]
Output: [0, 0, 1, 1, 1]
Explanation: The nums array in sorted order has 2 zeroes, 3 ones and zero twos.

```python
class Solution:
  def countNumbers(self,nums):
    count0=count1=count2=0

    for num in nums:
      if num==0:
        count0+=1
      elif num==1:
        count1+=1
      elif num==2:
        count2+=1

      index =0

      for i in range(count0):
        nums[index]=0
        index+=1
      for i in range(count1):
        nums[index]=1
        index+=1
      for i in range(count2):
        nums[index]=2
        index+=1

nums=[1,0,2,1,0]
obj =Solution()
obj.countNumbers(nums)
print(nums)
```

```python
#better solution
class Solution():
  def sortZeroOneTwo(self,nums):
    cnt0=cnt1=cnt2=0

    for num in nums:
      if num==0:
        cnt0+=1
      elif num==1:
        cnt1+=1
      elif num==2:
        cnt2+=1
    for i in range(cnt0):
      nums[i]=0
    for i in range(cnt0,cnt0+cnt1):
      nums[i]=1
    for i in range(cnt0+cnt1,len(nums)):
      nums[i]=2
    return nums

nums=[0,2,1,2,0,1]
sol=Solution()
sol.sortZeroOneTwo(nums)
print(nums)
```

```python
#Optimal Solution
class Solution():
  def sol(self,nums):
    low,mid,high=0,0,len(nums)-1
    while mid <=high:
      if nums[mid]==0:
        nums[low],nums[mid]=nums[mid],nums[low]
        low+=1
        mid+=1
      elif nums[mid]==1:
        mid+=1
      else:
        nums[mid],nums[high]=nums[high],nums[mid]
        high-=1

nums=[2,0,2,1,1,0]
obj=Solution()
obj.sol(nums)
print(nums)
```
